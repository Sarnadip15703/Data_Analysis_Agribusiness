## Agriculture dataset from kaggle 

```python
import pandas as pd

df = pd.read_csv("C:\\Users\\sarna\\Downloads\\agriculture_dataset.csv")
print(df.head())      
print(df.columns)     
```

```python
df['Yield_per_Acre'] = df['Yield(tons)'] / df['Farm_Area(acres)']
```

```python
result = df.groupby('Crop_Type')['Yield_per_Acre'].mean()
print(result.sort_values(ascending=False))
```

```python
result = df.groupby('Irrigation_Type')['Yield_per_Acre'].mean()
print(result.sort_values(ascending=False))
```

```python
df['Yield_per_1000m3_Water'] = df['Yield(tons)'] / (df['Water_Usage(cubic meters)'] / 1000)
```

```python
result = df.groupby('Irrigation_Type')['Yield_per_1000m3_Water'].mean()
print(result.sort_values(ascending=False))
```

```python
df['Yield_per_Ton_Fertilizer'] = df['Yield(tons)'] / df['Fertilizer_Used(tons)']
```

```python
result = df.groupby('Crop_Type')['Yield_per_Ton_Fertilizer'].mean()
print(result.sort_values(ascending=False))
```

```python
result = df.groupby('Soil_Type')['Yield_per_Acre'].mean()
print(result.sort_values(ascending=False))
```
